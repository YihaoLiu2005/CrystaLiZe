```python
def thermal_conductivity_ofhc(T: float, rrr: int) -> float:
    """Calculates OFHC Copper thermal conductivity k (W/m-K) at temperature T (Kelvin).
    Valid for 4 K <= T <= 300 K.
    """
    coeffs = {
        50:  {'a': 1.8743, 'b': -0.41538, 'c': -0.6018, 'd': 0.13294, 'e': 0.26426, 'f': -0.0219, 'g': -0.051276, 'h': 0.0014871, 'i': 0.003723},
        100: {'a': 2.2154, 'b': -0.47461, 'c': -0.88068, 'd': 0.13871, 'e': 0.29505, 'f': -0.02043, 'g': -0.04831, 'h': 0.001281, 'i': 0.003207},
        150: {'a': 2.3797, 'b': -0.4918, 'c': -0.98615, 'd': 0.13942, 'e': 0.30475, 'f': -0.019713, 'g': -0.046897, 'h': 0.0011969, 'i': 0.0029988},
        300: {'a': 1.357,  'b': 0.3981,   'c': 2.669,   'd': -0.1346, 'e': -0.6683, 'f': 0.01342,  'g': 0.05773,   'h': 0.0002147, 'i': 0.0},
        500: {'a': 2.8075, 'b': -0.54074, 'c': -1.2777, 'd': 0.15362, 'e': 0.36444, 'f': -0.02105, 'g': -0.051727, 'h': 0.0012226, 'i': 0.0030964}
    }
    
    if rrr not in coeffs:
        raise ValueError(f"RRR must be one of {list(coeffs.keys())}")
    if not (4 <= T <= 300):
        raise ValueError("Temperature T must be between 4 K and 300 K")

    p = coeffs[rrr]
    num = p['a'] + p['c']*(T**0.5) + p['e']*T + p['g']*(T**1.5) + p['i']*(T**2)
    den = 1 + p['b']*(T**0.5) + p['d']*T + p['f']*(T**1.5) + p['h']*(T**2)
    
    return 10**(num / den)

# Example Usage:
rrr = 50
Temp = 180
k_val = thermal_conductivity_ofhc(Temp, rrr)
print(f"k({Temp} K, RRR={rrr}) = {k_val:.1f} W/(m·K)")
```
